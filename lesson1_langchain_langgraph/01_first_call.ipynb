# Шаг 01 — первый вызов модели

Главная идея LangChain: модель — это объект с методом `.invoke()`.
OpenRouter говорит на протоколе OpenAI, поэтому берём обычный `ChatOpenAI` и меняем только `base_url`.

```python
import os

from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

load_dotenv()  # reads OPENROUTER_API_KEY and MODEL from .env

llm = ChatOpenAI(
    model=os.getenv("MODEL", "openai/gpt-4o-mini"),
    base_url="https://openrouter.ai/api/v1",   # OpenRouter speaks the OpenAI protocol
    api_key=os.getenv("OPENROUTER_API_KEY"),
)
```

```python
answer = llm.invoke("Объясни в двух предложениях, что такое AI-агент.")
print(answer.content)
```

Ответ — не строка, а объект `AIMessage`: текст, токены, метаданные провайдера.
